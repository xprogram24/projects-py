```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

file = 'messy_housing_data.csv'
df = pd.read_csv(file)
```

```python
df.head()
```

```python
null_count = df.isnull().sum()
null_count
```

```python
mean = df['bedrooms'].astype(float).mean()
df['bedrooms'] = df['bedrooms'].fillna(mean)
```

```python
df.info()
```

```python
df['price'] = df['price'].str.replace("$","").str.replace(",",'').str.replace("?",'')
```

```python
plt.figure(figsize=(9, 4.5))
price = df['price']
year = df['year_built']

# Works cleanly on 50,000+ rows out of the box
df['bedrooms'].plot(kind='hist', bins=30, edgecolor='black', figsize=(8, 4))
plt.title('price of houses')
plt.show()
```

```python
df['conditionplt'] = df.groupby('condition')['price']
df['conditionplt'].plot(kind='pie')
plt.show()
```

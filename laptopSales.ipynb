```python
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup

# Raw HTML from web page
html_data = """
<div id="products">
    <h2>Tech Deals</h2>
    <table class="data-table">
        <thead>
            <tr>
                <th>Device</th>
                <th>Price_USD</th>
                <th>Units_Sold</th>
            </tr>
        </thead>
        <tbody>
            <tr>
                <td class="name">ThinkPad P14s</td>
                <td class="price">$1,450.00</td>
                <td class="units">120</td>
            </tr>
            <tr>
                <td class="name">MacBook Pro 14</td>
                <td class="price">$1,999.00</td>
                <td class="units">85</td>
            </tr>
            <tr>
                <td class="name">Dell XPS 15</td>
                <td class="price">$1,200.00</td>
                <td class="units">N/A</td>
            </tr>
            <tr>
                <td class="name">Asus ROG Zephyrus</td>
                <td class="price">$1,650.00</td>
                <td class="units">210</td>
            </tr>
        </tbody>
    </table>
</div>
"""
```

```python
soup = BeautifulSoup(html_data, 'html.parser')

table_body = soup.find('tbody')
rows =  table_body.find_all('tr')

scraped_data = []

for row in rows:
    cells = row.find_all('td')


    #extract the clean text inside each td
    device = cells[0].text.strip()
    price = cells[1].text.strip()
    units = cells[2].text.strip()

    scraped_data.append({
        'device':device,
        'price (usd)':price,
        'units':units
    })


    
```

```python
#load into a pandas dataframe
df = pd.DataFrame(scraped_data)
print("--- Raw Scraped DataFrame ---")
print(df)
```

```python
df['price (usd)'] = df['price (usd)'].str.replace('$','').str.replace(',','').astype(float)
```

```python
print(df)
```

```python
df['units'] = df['units'].replace('N/A', np.nan).astype(float)

df['Revenue_USD'] = df['price (usd)'] * df['units']
```

```python
print("--- Cleaned DataFrame ---")
print(df)
```

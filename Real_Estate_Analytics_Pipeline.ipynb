```python
import pandas as pd
import numpy as np
from bs4 import BeautifulSoup

html_data = """
<div class="listings">
    <h2>Available Properties</h2>
    <table id="property_table">
        <thead>
            <tr>
                <th>Location</th>
                <th>Price_Per_SqFt</th>
                <th>Total_SqFt</th>
            </tr>
        </thead>
        <tbody>
            <tr>
                <td class="loc">Lekki Phase 1</td>
                <td class="price">$150.00</td>
                <td class="sqft">2,500</td>
            </tr>
            <tr>
                <td class="loc">Ikoyi</td>
                <td class="price">$320.00</td>
                <td class="sqft">1,800</td>
            </tr>
            <tr>
                <td class="loc">Victoria Island</td>
                <td class="price">$280.00</td>
                <td class="sqft">N/A</td>
            </tr>
            <tr>
                <td class="loc">Ikeja GRA</td>
                <td class="price">$110.00</td>
                <td class="sqft">3,100</td>
            </tr>
        </tbody>
    </table>
</div>
"""
```

```python
soup = BeautifulSoup(html_data, 'html.parser')
table_body = soup.find('tbody')
rows = table_body.find_all('tr'
)
data = []

for row in rows:
    cells = row.find_all('td')
    location = cells[0].text.strip()
    price_sqft = cells[1].text.strip()
    total_sqft = cells[2].text.strip()

    data.append({
        'location': location,
        'price_per_sqft': price_sqft,
        'total_sqft': total_sqft
    })

df = pd.DataFrame(data)

```

```python
print(df)
```

```python
#Clean Price_Per_SqFt: Remove the $ sign and convert to float.
df['price_per_sqft'] = df['price_per_sqft'].str.replace('$','').astype(float)
```

```python
#Clean Total_SqFt: Remove the comma ,, replace 'N/A' with np.nan, and convert to float
df['total_sqft'] = df['total_sqft'].str.replace(',','').replace('N/A', np.nan).astype(float)
```

```python
#Calculate Total_Price: Create a new column df['Total_Price'] by multiplying Price_Per_SqFt by Total_SqFt.
df['Total_Price'] = df['price_per_sqft'] * df['total_sqft']
```

```python
print("--- Cleaned DataFrame ---")
print(df)
```

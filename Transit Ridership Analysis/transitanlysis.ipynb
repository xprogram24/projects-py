```python
import os
import pandas as pd 
csv_name =  "cta-ridership-daily-boarding-totals.csv"
file_path = os.path.join(path,csv_name)

#read the csv
ridership_df = pd.read_csv(file_path)

#structual audit
print(f"shape :{ridership_df.shape}")
print("\nColumn Data Types & Non-Nulls:")
print(ridership_df.info())
print(f'data description {ridership_df.describe()}')
```

```python
print(ridership_df.head())
print(ridership_df.tail())
```

```python
#parse data column
ridership_df['service_date'] = pd.to_datetime(ridership_df['service_date'])

#check full date range
min_date  = ridership_df['service_date'].min()
max_date =  ridership_df['service_date'].max()

print(f"Dataset covers: {min_date.strftime('%Y-%m-%d')} to {max_date.strftime('%Y-%m-%d')}")

#filter to a clean five year window
start_window = '2015-01-01'
end_window = '2019-12-31'

filtered_ridership = ridership_df[
    (ridership_df['service_date'] >= start_window) & (ridership_df['service_date'] <= end_window)
].copy()


print(f"Filtered subset shape: {filtered_ridership.shape}")
print(filtered_ridership.head())
```

```python
import requests

# Open-Meteo Historical Archive API (Free, no API key required)
weather_url = "https://archive-api.open-meteo.com/v1/archive"

params = {
    "latitude": 41.8781,
    "longitude": -87.6298,
    "start_date": "2015-01-01",
    "end_date": "2019-12-31",
    "daily": [
        "temperature_2m_mean",
        "temperature_2m_max",
        "temperature_2m_min",
        "precipitation_sum",
        "rain_sum",
        "snowfall_sum",
        "wind_speed_10m_max"
    ],
    "timezone": "America/Chicago"
}

response = requests.get(weather_url, params=params)
weather_json = response.json()

# Convert daily JSON to DataFrame
weather_df = pd.DataFrame(weather_json["daily"])
weather_df["time"] = pd.to_datetime(weather_df["time"])
weather_df.rename(columns={"time": "service_date"}, inplace=True)

print("Weather Data Shape:", weather_df.shape)
print(weather_df.head())
```

```python
# Perform inner join on the service_date column
df_merged = pd.merge(
    filtered_ridership,weather_df, on='service_date',how='inner'
)

print(f"mergered dataset shape {df_merged.shape}")
print(df_merged.info())
print(df_merged.head())
```

```python
# 1. Temporal Features
df_merged["day_of_week"] = df_merged["service_date"].dt.dayofweek
df_merged["month"] = df_merged["service_date"].dt.month
df_merged["year"] = df_merged["service_date"].dt.year
df_merged["is_weekday"] = (df_merged["day_type"] == "W").astype(int)

# 2. Meteorological Threshold Features
df_merged["is_rainy"] = (df_merged["rain_sum"] > 0.5).astype(int)
df_merged["is_snowy"] = (df_merged["snowfall_sum"] > 0.0).astype(int)
df_merged["heavy_rain"] = (df_merged["precipitation_sum"] >= 10.0).astype(int)

# 3. Temperature Stress Features (Baseline = 20°C / 68°F)
df_merged["temp_deviation"] = (df_merged["temperature_2m_mean"] - 20.0).abs()
df_merged["is_freezing"] = (df_merged["temperature_2m_mean"] < 0.0).astype(int)

# 4. Quick Sanity Check
print("Engineered DataFrame Columns:")
print(df_merged.columns.tolist())
print("\nSample Engineered Features:")
print(df_merged[[
    "service_date", "day_type", "total_rides", 
    "temperature_2m_mean", "temp_deviation", 
    "rain_sum", "is_rainy", "snowfall_sum", "is_snowy"
]].head())
```

```python
# 1. Compare average ridership on Dry vs Rainy days across Day Types
rain_impact = df_merged.groupby(["day_type", "is_rainy"])[
    ["bus", "rail_boardings", "total_rides"]
].mean().round(0)

print("Average Daily Ridership: Dry (0) vs Rainy (1)")
print(rain_impact)

# 2. Calculate percentage ridership drop on rainy days per day type
print("\nPercentage Change on Rainy Days:")
for day in ["W", "A", "U"]:
    dry_val = rain_impact.loc[(day, 0), "total_rides"]
    rain_val = rain_impact.loc[(day, 1), "total_rides"]
    pct_change = ((rain_val - dry_val) / dry_val) * 100
    day_label = {"W": "Weekday", "A": "Saturday", "U": "Sunday/Holiday"}[day]
    print(f"- {day_label} ({day}): {pct_change:.2f}%")
```

```python
# Select numeric features for correlation audit
corr_features = [
    "bus", 
    "rail_boardings", 
    "total_rides", 
    "temperature_2m_mean", 
    "temp_deviation", 
    "precipitation_sum", 
    "rain_sum", 
    "snowfall_sum", 
    "wind_speed_10m_max"
]

corr_matrix = df_merged[corr_features].corr().round(3)

print("Correlation with Total Rides:")
print(corr_matrix["total_rides"].sort_values(ascending=False))
```

```python
import statsmodels.api as sm

# 1. Create dummy variables for day_type (drops one baseline category to prevent the dummy variable trap)
# Baseline category will be 'W' (Weekday)
day_dummies = pd.get_dummies(df_merged["day_type"], prefix="day", drop_first=True, dtype=int)

# 2. Define Explanatory Features (X)
# Controlling for: Saturday (A), Sunday (U), Rain, Snow, Extreme Temp, Wind Speed
X_features = pd.concat([
    day_dummies,
    df_merged[[
        "rain_sum",
        "snowfall_sum",
        "temp_deviation",
        "wind_speed_10m_max"
    ]]
], axis=1)

# 3. Add constant intercept term (Beta 0)
X = sm.add_constant(X_features)

# 4. Define Target Variable (y)
y = df_merged["total_rides"]

# 5. Fit the OLS Model
model = sm.OLS(y, X).fit()

# 6. Display Statistical Summary
print(model.summary())
```

```python
# 1. Create the interaction term: Rain x Weekday
X_features["rain_x_weekday"] = X_features["rain_sum"] * X_features["day_W"]

# 2. Add constant
X_interaction = sm.add_constant(X_features)

# 3. Fit Model with Interaction
model_interaction = sm.OLS(y, X_interaction).fit()

# 4. Print the key coefficients
print(model_interaction.summary().tables[1])
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

# 1. Use the original feature set (without interaction)
X_clean = sm.add_constant(pd.concat([
    day_dummies,
    df_merged[[
        "rain_sum",
        "snowfall_sum",
        "temp_deviation",
        "wind_speed_10m_max"
    ]]
], axis=1))

# 2. Calculate VIF for each feature
vif_data = pd.DataFrame()
vif_data["Feature"] = X_clean.columns
vif_data["VIF"] = [variance_inflation_factor(X_clean.values, i) for i in range(X_clean.shape[1])]

print("Multicollinearity Audit (VIF):")
print(vif_data.sort_values(by="VIF", ascending=False))
```

```python
import plotly.express as px
import plotly.graph_objects as go

# Ensure predictions are generated from your fitted model
df_merged["predicted_rides"] = model.predict(X)
```

```python
# Model Fit: Actual vs. Predicted Daily Ridership ($R^2 = 0.886$)
#This plot demonstrates how accurately the model captures real passenger swings over the 5-year evaluation window.
fig_fit = px.scatter(
    df_merged,
    x="total_rides",
    y="predicted_rides",
    color="day_type",
    color_discrete_map={"W": "#38bdf8", "A": "#34d399", "U": "#f43f5e"},
    labels={
        "total_rides": "Actual Total Boardings",
        "predicted_rides": "OLS Model Predicted Boardings",
        "day_type": "Day Type"
    },
    title="<b>Actual vs. Model Predicted Transit Ridership</b>",
    template="plotly_dark",
    opacity=0.65
)

# Add 45-degree reference line for a perfect model fit
min_val = min(df_merged["total_rides"].min(), df_merged["predicted_rides"].min())
max_val = max(df_merged["total_rides"].max(), df_merged["predicted_rides"].max())

fig_fit.add_trace(
    go.Scatter(
        x=[min_val, max_val],
        y=[min_val, max_val],
        mode="lines",
        name="Perfect Fit (y = x)",
        line=dict(color="#e2e8f0", dash="dash", width=2)
    )
)

fig_fit.update_layout(
    font=dict(family="monospace"),
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)

fig_fit.show()
```

```python
#This visualizes the inverted parabolic drop as temperatures move toward freezing extremes, stratified by commuter baseline.
fig_temp = px.scatter(
    df_merged,
    x="temperature_2m_mean",
    y="total_rides",
    color="day_type",
    trendline="lowess",
    color_discrete_map={"W": "#38bdf8", "A": "#34d399", "U": "#f43f5e"},
    labels={
        "temperature_2m_mean": "Daily Mean Temperature (°C)",
        "total_rides": "Total Daily Boardings",
        "day_type": "Day Type"
    },
    title="<b>Transit Ridership Sensitivity to Temperature Extremes</b>",
    template="plotly_dark",
    opacity=0.5
)

fig_temp.update_layout(
    font=dict(family="monospace"),
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)

fig_temp.show()
```

```python
#Precipitation Impact Breakdown: Dry vs. Rainy Days
#A grouped box plot showing the exact distribution shift in daily ridership during wet conditions.
# Map readable labels for visualization
df_merged["weather_condition"] = df_merged["is_rainy"].map({0: "Dry Day", 1: "Rainy Day (>0.5mm)"})
df_merged["day_label"] = df_merged["day_type"].map({"W": "Weekday", "A": "Saturday", "U": "Sunday/Holiday"})

fig_rain = px.box(
    df_merged,
    x="day_label",
    y="total_rides",
    color="weather_condition",
    color_discrete_map={"Dry Day": "#0ea5e9", "Rainy Day (>0.5mm)": "#f59e0b"},
    category_orders={"day_label": ["Weekday", "Saturday", "Sunday/Holiday"]},
    labels={
        "day_label": "Day Category",
        "total_rides": "Total Daily Boardings",
        "weather_condition": "Precipitation State"
    },
    title="<b>Impact of Rainfall on Transit Boardings by Day Category</b>",
    template="plotly_dark"
)

fig_rain.update_layout(
    boxmode="group",
    font=dict(family="monospace"),
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)

fig_rain.show()
```
